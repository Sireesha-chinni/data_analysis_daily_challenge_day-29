# Coding Challenge: Student Marks Analyzer (Dictionary)
    
    
    
Problem Statement


You are given student marks in different subjects.
Use a dictionary to:
    
    
1. Store student data

```python
students = {
"Asha": {"Math": 78, "Science": 85, "English": 67},
"Ravi": {"Math": 35, "Science": 40, "English": 50},
"John": {"Math": 90, "Science": 92, "English": 88}
}
```

2. Calculate:

o Total marks
    
o Average marks

```python
totals = {}
averages = {}
for name, marks in students.items():
    totals[name] = sum(marks.values())
    averages[name] = round(totals[name] / len(marks), 2)
    print(f"{name}: Total = {totals[name]}, Average = {averages[name]}")
```

3. Find:
o Top scorer
o Failed students (marks < 40 in any subject)

```python
top_scorer = max(totals, key=totals.get)
print("Top scorer:", top_scorer, "with", totals[top_scorer], "marks")
```

```python
failed = []
for name, marks in students.items():
    for subject, score in marks.items():
        if score < 40:
            failed.append(name)
            break
print("Failed students:", failed)
```
